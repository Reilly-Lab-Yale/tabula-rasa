This notebook abstracts a bounds object from the cohen retina sMPRA data.

```python
import scMPRAforge as scm
%load_ext autoreload
%autoreload 2

from pathlib import Path
```

```python
from dask.distributed import Client, LocalCluster
cluster=LocalCluster(memory_limit='8GB')
client=Client(cluster)
```

Load cohen ortho object

```python
data_root=Path("/nfs/roberts/project/pi_skr2/shared/tabula_data")
primordial=scm.ortho.load(client,path=data_root,name="cohen_ortho")
```

```python
cohen_bounds=scm.Bounds.from_ortho(primordial)
```

```python
cohen_bounds.transfection_model.plot()
```

```python
cohen_bounds.library_model.plot()
```

```python
cohen_bounds.to_tgz("scMPRAforge/presets/cohen_bounds.tgz")
```

```python
client.close()
cluster.close()
```
