```python
percent_collision = {
    "Zhao et al.":    0.11689209661257542,
    "Lalanne et al.": 0.05304031883524153,
    "Yin et al.":     7.630123998189407,
}
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# keep SVG text as <text> elements so labels stay editable downstream
plt.rcParams["svg.fonttype"] = "none"

collision_df = (
    pd.DataFrame(
        {"method": list(percent_collision.keys()), "collision_rate": list(percent_collision.values())}
    )
    .dropna(subset=["collision_rate"])
    .sort_values("collision_rate", ascending=False)
)

sns.set_theme(style="whitegrid")
fig, ax = plt.subplots(figsize=(4, 3))
sns.barplot(data=collision_df, x="method", y="collision_rate", ax=ax, color="#4C78A8")
ax.set_xlabel("")
ax.set_ylabel("")
ax.set_title("Estimated collision rate (fraction of total transfection events)")
ax.set_ylim(0, max(collision_df["collision_rate"].max() * 1.15, 0.01))
for container in ax.containers:
    ax.bar_label(container, fmt="%.1f%%", labels=[f"{v:.3f}%" for v in collision_df["collision_rate"]])
fig.tight_layout()
out_svg = "./estimated_percent_conflict_collision_rate.svg"
fig.savefig(out_svg, format="svg", bbox_inches="tight")
out_svg
```
