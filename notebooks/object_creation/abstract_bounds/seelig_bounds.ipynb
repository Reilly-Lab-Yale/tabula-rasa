This notebook abstracts a bounds object from the Seelig scMPRA data (consider_missing=True ZINB ortho).

```python
import scMPRAforge as scm
%load_ext autoreload
%autoreload 2

from pathlib import Path
```

```python
from dask.distributed import Client, LocalCluster
cluster = LocalCluster(memory_limit='8GB')
client = Client(cluster)
```

Load seelig ortho object

```python
data_root = Path("/nfs/roberts/project/pi_skr2/shared/tabula_data")
primordial = scm.ortho.load(client, path=data_root / "seelig", name="seelig_ortho_20260320")
```

```python
seelig_bounds = scm.Bounds.from_ortho(primordial)
```

```python
seelig_bounds.transfection_model.plot()
```

```python
seelig_bounds.library_model.plot()
```

```python
seelig_bounds.to_tgz("scMPRAforge/presets/seelig_bounds.tgz")
```

```python
client.close()
cluster.close()
```
