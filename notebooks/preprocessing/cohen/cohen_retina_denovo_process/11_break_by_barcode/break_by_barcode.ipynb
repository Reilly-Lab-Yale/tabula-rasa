# Initial loading

Importing and setting up global vars...

```python
import pandas as pd
import os
from umi_tools import UMIClusterer

DATA_ROOT="/home/mcn26/palmer_scratch/raw_recap/cohen_retina"
umi_len=12
pBC_len=8
```

Load MPRA barcode data into a table.

```python
def break_up_mpra_count(file_name_in):
    # Define expected barcode lengths
    rBC_len = 24
    
    # List to store parsed data
    data = []
    
    # Read file and process lines
    with open(os.path.join(DATA_ROOT, file_name_in)) as fin:
        for line in fin:
            line = line.strip()
            
            # Break off number of occurrences
            parts = line.split(" ")
            assert len(parts) == 2, "Error: multiple spaces"
            num_occurrences = int(parts[0])
            
            # Break off cell barcode
            subparts = parts[1].split("_")
            assert len(subparts) == 2, "Error: multiple underscores"
            cell_barcode = subparts[0]
            assert len(cell_barcode) == 16, "Error: malformed cell barcode"
            
            # Break out remaining barcodes
            barcode_seq = subparts[1]
            assert len(barcode_seq) == 44, "Error: malformed remaining barcode"
            
            umi = barcode_seq[:umi_len]
            pBC = barcode_seq[umi_len:umi_len + pBC_len]
            rBC = barcode_seq[umi_len + pBC_len:]
            
            # Append parsed data
            data.append([num_occurrences, cell_barcode, umi, pBC, rBC])
    
    # Create DataFrame
    df = pd.DataFrame(data, columns=["num_occurrences", "cell_barcode", "umi", "pBC", "rBC"])
    
    return df

# Example usage
mpra_r1 = break_up_mpra_count("counts/retina_mpra_r1.count")
mpra_r2 = break_up_mpra_count("counts/retina_mpra_r2.count")
```

Let's load the u6 data..

```python
def break_up_u6_count(filename):
    #format is
    #leading whitespace, num occurances, space, 16 bp cell-barcode, 12bp UMI, 8 bp pBC
    
    # List to store parsed data
    data = []

    with open(f"{DATA_ROOT}/{filename}") as fin:
        for line in fin:
            original_line=line

            line=line.strip()
            
            #break off number of occurrences
            line=line.split(" ")
            assert len(line) == 2, "err, multiple spaces"
            num_occurrences=int(line[0])

            #break off cell barcode
            line=line[1].split("_")
            assert len(line) == 2, "err, multiple underscore"
            cell_barcode=line[0]
            assert len(cell_barcode)==16, "err, malformed cell barcode"

            #break out remaining barcodes
            line=line[1]
            assert len(line)==20, "err, malformed remaining barcode"
            
            umi=line[:umi_len]
            pBC=line[umi_len:]
            
            data.append([num_occurrences, cell_barcode, umi, pBC])

    
    return pd.DataFrame(data,columns=["num_occurrences","cell_barcode","umi","pBC"])



u6_r1=break_up_u6_count("counts/retina_u6_r1.count")
u6_r2=break_up_u6_count("counts/retina_u6_r2.count")
```

# Filtering main MPRA data

We will merge the cell-type info in. Since this data has gone through filtering, an inner join will filter the MPRA data, dropping bad cells.

Let's stack the MPRA replicates:

```python
mpra_r1["replicate"]="1"
mpra_r2["replicate"]="2"
all_mpra=pd.concat([mpra_r1,mpra_r2])
```

```python
all_mpra
```

I want to do inner joins of mpra data and cell-type annotations, but I want to make sure there will be no duplicated entries. So I will make sure all cell-barcode, replicate combinations are unique in the cell type annotation df. 

```python
total_celltype_rows=len(cell_types)
unique_celltype_cbc=cell_types.groupby(["cell_barcode","replicate"]).ngroups
assert total_celltype_rows==unique_celltype_cbc
```

Perfect. Let's do the intersection:

```python
joincols=["cell_barcode","replicate"]

all_mpra=all_mpra.astype({col:str for col in joincols})
cell_types=cell_types.astype({col:str for col in joincols})

inner_mpra=all_mpra.merge(cell_types,on=joincols,how="inner")
inner_mpra
```

```python
print(len(inner_mpra))
print(len(all_mpra))
print(len(cell_types))
print()
print(f"retained={len(inner_mpra)/len(all_mpra)*100}%")
```

Let's check if complementing or reverse-complementing bases 8 & 9 would increase retained % (since this is a feature of some 10x cell-barcodes...):

```python
#some helper functions 
comp = str.maketrans("ACGTacgt", "TGCAtgca")

def comp_mid(nt):
    return nt[0:7]+ nt[7:9].translate(comp) +nt[9:16]

def revcomp_mid(nt):
    return nt[0:7]+ nt[7:9].translate(comp)[::-1] +nt[9:16]

#quick tests
assert "GATCgatcg".translate(comp)=="CTAGctagc"
assert comp_mid("AAAAAAATCAAAAAAA")=="AAAAAAAAGAAAAAAA"
assert revcomp_mid("AAAAAAATCAAAAAAA")=="AAAAAAAGAAAAAAAA"
```

```python
all_mpra["rc_cbc"]=all_mpra["cell_barcode"].apply(revcomp_mid)
all_mpra["c_cbc"]=all_mpra["cell_barcode"].apply(comp_mid)
```

```python
#spot-checking..
all_mpra
```

```python
rc_height=len(all_mpra.merge(cell_types,left_on=["rc_cbc","replicate"],right_on=["cell_barcode","replicate"],how="inner"))
print(f"retained rc={rc_height/len(all_mpra)*100}%")

c_height=len(all_mpra.merge(cell_types,left_on=["c_cbc","replicate"],right_on=["cell_barcode","replicate"],how="inner"))
print(f"retained c={c_height/len(all_mpra)*100}%")
```

Clearly the join `inner_mpra` was best... 68% remaining is not so bad a loss. I will go ahead with this for now, and implement a more sophisticated system using the umi-tools API later.

# Filtering U6 data

First thing we need to do is figure out how the barcodes are to be merged in...

```python
all_u6["rc_cbc"]=all_u6["cell_barcode"].apply(revcomp_mid)
all_u6["c_cbc"]=all_u6["cell_barcode"].apply(comp_mid)

inner_u6=all_u6.merge(cell_types,on=["cell_barcode","replicate"],how="inner")
rc_inner_u6=all_u6.merge(cell_types,left_on=["rc_cbc","replicate"],right_on=["cell_barcode","replicate"],how="inner")
c_inner_u6=all_u6.merge(cell_types,left_on=["c_cbc","replicate"],right_on=["cell_barcode","replicate"],how="inner")

print(f"retained={len(inner_u6)/len(all_u6)*100}%")
print(f"retained rc={len(rc_inner_u6)/len(all_u6)*100}%")
print(f"retained c={len(c_inner_u6)/len(all_u6)*100}%")
```

Plainly the complement is correct! Let's clean up `c_inner_u6` and proceed...

```python
u6=c_inner_u6.drop(["cell_barcode_x","cell_barcode_y","rc_cbc"],axis=1)
u6=u6.rename({"c_cbc":"cell_barcode"},axis=1)
```
