```python
%%writefile simple_fit.py
print("[+] Imports...",flush=True)
import sys
import os
import time
package_path = os.path.abspath("../..")  
sys.path.insert(0, package_path)
#The path will be managed by conda or whatever on release, but this is fine for now.
import scMPRAforge as scm
import pickle

data_root="/gpfs/gibbs/pi/reilly/tabula_data"
print(f"[+] Loading data... {time.time()}",flush=True)
sh=scm.load_scMPRA_data(f"{data_root}/shendure/shendure_counts_grouped.txt")

print(f"[+] Fitting... {time.time()}",flush=True)
sh_model=scm.experiment_model(sh,'simple_cell_and_cre')
sh_model.fit()

print(f"[+] Dumping... {time.time()}")
with open(f"{data_root}/shendure/forge_models/simple_cell_and_cre.pkl", "wb") as f:
    pickle.dump(sh_model, f)
```
