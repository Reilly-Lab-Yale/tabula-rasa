Converts the ground truth dataframe to a better format

```python
import pandas as pd
```

```python
ground_truth_mu=pd.read_csv("synthetic_ground_truth.tsv",sep="\t")
#fix col names
ground_truth_mu=ground_truth_mu.rename({'CRE':'cre_id','mean':'true_mu','Cell-type':'cell_type'},axis=1)
#set reference CRE
ground_truth_mu.loc[
    ground_truth_mu["cre_id"].isin(["nobody","weak"]),
    "cre_id"
]="reference"
#set reference cell-type
ground_truth_mu.loc[
    ground_truth_mu["cell_type"]=="liver",
    "cell_type"
]="reference"
#set index
ground_truth_mu=ground_truth_mu.set_index(['cell_type','cre_id'])
#save

#display
ground_truth_mu
```

```python
ground_truth_mu.to_csv("synthetic_ground_truth_carp.tsv",sep="\t")
```
