```python
import scMPRAforge as scm
import pandas as pd
from formulaic import Formula
```

```python
two=pd.DataFrame({'ct':['a','a','a','b','b','b']})
three=pd.DataFrame({'ct':['a','c','a','b','c','b']})

two_design=Formula('C(ct)').get_model_matrix(two,output='pandas')
three_design=Formula('C(ct)').get_model_matrix(three,output='pandas')

multi=pd.DataFrame({'ct':['a','a','a','b','b','b'],'foo':['1','2','3','1','2','3']})
multi_design=Formula('C(ct)+C(foo)').get_model_matrix(multi,output='pandas')
multi_no_intercept=Formula('C(ct)+C(foo)-1').get_model_matrix(multi,output='pandas')
```
