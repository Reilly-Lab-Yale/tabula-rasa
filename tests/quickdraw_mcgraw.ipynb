```python
import dask.array as da
import numpy as np
import time

# Parameters
shape = (778_015,)         # total number of draws
approx_chunk_size = 1_000_000 # approximate size per chunk
chunks = (approx_chunk_size,)

# Create random n and p arrays
rng = da.random.RandomState(42)
n = rng.uniform(1, 10, size=shape, chunks=chunks)
p = rng.uniform(0.1, 0.9, size=shape, chunks=chunks)

# Time the negative binomial draw
start = time.time()
samples = da.random.negative_binomial(n=n, p=p, size=shape, chunks=chunks)
_ = samples.mean().compute()  # trigger computation
end = time.time()

print(f"Sampling {shape[0]:,} values took {end - start:.2f} seconds.")
```
